# ARC AI2 Reasoning Challenge

https://huggingface.co/datasets/allenai/ai2_arc (also https://www.kaggle.com/datasets/jeromeblanchet/arc-ai2-reasoning-challenge?select=ARC-Challenge-Test.csv)

````python
arc_easy = load_dataset("allenai/ai2_arc", "ARC-Easy", revision="210d026faf9955653af8916fad021475a3f00453")[
    "train"].to_pandas()
arc_easy
````

````python
arc_easy.iloc[4]["answerKey"]
````

````python
# Transformation logic
arc_easy.apply(
    lambda row: pd.Series({
        'question': row['question'],
        'choices': row['choices']['text'],
        'correct_answer': list(row['choices']['label']).index(row['answerKey'])
    }),
    axis=1
)
````

````python
arc_challenge = load_dataset("allenai/ai2_arc", "ARC-Challenge", revision="210d026faf9955653af8916fad021475a3f00453")[
    "train"].to_pandas()
arc_challenge
````
